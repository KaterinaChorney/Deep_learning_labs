Навчання фінальної моделі

```python
import time
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils
import tf2onnx
import onnx


(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

x_train_flat = x_train.reshape(60000, 784) / 255.0
x_test_flat = x_test.reshape(10000, 784) / 255.0

y_train_cat = utils.to_categorical(y_train, 10)
y_test_cat = utils.to_categorical(y_test, 10)

final_model = Sequential([
    Dense(800, input_dim=784, activation="relu"),
    Dense(900, activation="relu"), 
    Dense(10, activation="softmax")
])

final_model.compile(
    loss="categorical_crossentropy", 
    optimizer="SGD", 
    metrics=["accuracy"]
)

print("Навчання фінальної моделі:")
t_start = time.time()
final_history = final_model.fit(
    x_train_flat, y_train_cat, 
    batch_size=50, 
    epochs=125, 
    validation_split=0.2, 
    verbose=1
)
t_end = time.time()

scores = final_model.evaluate(x_test_flat, y_test_cat, verbose=0)
print(f"\nТочність фінальної моделі на тесті: {round(scores[1] * 100, 2)}%")
print(f"Час навчання: {round((t_end - t_start) / 60, 2)} хв")
```

Збереження моделі

```python
final_model.save('fashion_mnist_dense.h5')
print("\nМодель успішно збережено у файл 'fashion_mnist_dense.h5'")

final_model.save('fashion_mnist_dense.keras')
print("Модель успішно збережено у файл 'fashion_mnist_dense.keras'")

try:
    @tf.function
    def model_func(x):
        return final_model(x)

    onnx_model, _ = tf2onnx.convert.from_function(
        model_func, 
        input_signature=[tf.TensorSpec([None, 784], tf.float32, name='dense_input')]
    )
    onnx.save(onnx_model, "fashion_mnist_dense.onnx")
    print("Модель успішно конвертовано та збережено у файл 'fashion_mnist_dense.onnx'")
except Exception as e:
    print(f"Помилка при збереженні в ONNX: {e}")
```
