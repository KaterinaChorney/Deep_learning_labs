Завантаження та підготовка даних

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from keras.datasets import cifar10
from keras.utils import to_categorical
from keras.models import Sequential, model_from_json
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
import os
from tensorflow.keras.preprocessing.image import load_img, img_to_array

(X_train, y_train), (X_test, y_test) = cifar10.load_data()
print(f"Розмірність навчальних даних: {X_train.shape}")
print(f"Розмірність тестових даних: {X_test.shape}")

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
print(f"Приклад one-hot мітки: {y_train_cat[0]}")
```

Створення, навчання та оцінка моделі

```python
model = Sequential()

model.add(Conv2D(32, (3, 3), padding="same", activation="relu", input_shape=(32, 32, 3)))
model.add(Conv2D(32, (3, 3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.25))

model.add(Conv2D(64, (3, 3), padding="same", activation="relu"))
model.add(Conv2D(64, (3, 3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Dropout(0.25))

model.add(Flatten())
model.add(Dense(512, activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(10, activation="softmax"))

model.summary()

model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])

history = model.fit(X_train, y_train_cat, 
                    batch_size=64, 
                    epochs=15, 
                    validation_data=(X_test, y_test_cat), 
                    verbose=1)

scores = model.evaluate(X_test, y_test_cat, verbose=0)
print(f"Втрати на тестових даних: {scores[0]:.4f}")
print(f"Точність на тестових даних: {scores[1]:.4f}")

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Навчальна вибірка")
plt.plot(history.history["val_accuracy"], label="Тестова вибірка")
plt.title("Точність моделі")
plt.ylabel("Точність")
plt.xlabel("Епоха")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Навчальна вибірка")
plt.plot(history.history["val_loss"], label="Тестова вибірка")
plt.title("Втрати моделі")
plt.ylabel("Втрати")
plt.xlabel("Епоха")
plt.legend()
plt.show()

model_json = model.to_json()
with open("cifar10_model.json", "w") as json_file:
    json_file.write(model_json)
model.save_weights("cifar10_model_weights.weights.h5")
print("Модель збережена.")
```

Розпізнавання зображень

```python
classes = ["літак", "автомобіль", "птах", "кіт", "олень", "собака", "жаба", "кінь", "корабель", "вантажівка"]

fig, axes = plt.subplots(3, 5, figsize=(15, 9))
axes = axes.flatten()

predictions = model.predict(X_test[:15])

print("Розпізнавання тестових зображень:")

for i, ax in enumerate(axes):
    ax.imshow(X_test[i])
    pred_class = classes[np.argmax(predictions[i])]
    true_class = classes[np.argmax(y_test_cat[i])]
    title_color = "green" if pred_class == true_class else "red"
    ax.set_title(f"Висновок: {pred_class}\nСправжнє: {true_class}", color=title_color)
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Розпізнавання власних зображень:")
folder_path = "my_images"
my_images = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.jpg')]

def predict_custom_image(image_path):
    img = load_img(image_path, target_size=(32, 32))
    plt.figure(figsize=(2,2))
    plt.imshow(img)
    plt.axis("off")
    plt.show()
    
    x = img_to_array(img)
    x = x.astype("float32") / 255.0
    x = np.expand_dims(x, axis=0)
    
    prediction = model.predict(x)
    pred_class_idx = np.argmax(prediction)
    print(f"Зображення {image_path}: {classes[pred_class_idx]} (Імовірність: {np.max(prediction):.2f})\n")

for filename in my_images:
    predict_custom_image(filename)
```
